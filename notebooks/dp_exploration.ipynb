## 6 · Budget Composition

```python
budget = PrivacyBudget(total_epsilon=3.0, total_delta=1e-5)
for lbl, eps in [('Count A', 0.5), ('Sum B', 1.0), ('Hist C', 0.5), ('Count D', 1.5), ('Count E', 0.5)]:
    ok = budget.spend(eps, label=lbl)
    print(f"{'✓' if ok else '✗'} {lbl} (ε={eps})  → cumulative ε = {budget.spent_epsilon:.2f}")

p = plot_budget_composition(Path('.'))
Image(p)
```
